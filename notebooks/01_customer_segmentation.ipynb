```python
X = df[["Annual Income (k$)","Spending Score (1-100)"]]
```

```python
X.head()
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled[:5]
```

### Find the Optimal K — Elbow Method

```python
inertia = []

k_range = range(1, 11)

for k in k_range:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(X_scaled)

    inertia.append(kmeans.inertia_)
```

### Plot Elbow Curve

```python
plt.figure(figsize=(8, 5))
plt.plot(
    k_range,
    inertia,
    marker="o"
)
plt.title("Elbow Method")
plt.xlabel("Number of clusters (k)")
plt.ylabel("Inertia")
plt.xticks(k_range)
plt.show()
```

### Silhouette Score

```python
silhouette_scores = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels
    )

    silhouette_scores.append(score)
```

### Plot Silhouette Scores

```python
plt.figure(figsize=(8, 5))
plt.plot(
    range(2, 11),
    silhouette_scores,
    marker="o"
)

plt.title("Silhouette Score by Number of Clusters")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.xticks(range(2, 11))
plt.show()
```

```python
optimal_k = 5

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_scaled)
```

```python
df['Cluster'] = clusters

df.head()
```

### Visualize the Clusters

```python
plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=df,
    x="Annual Income (k$)",
    y="Spending Score (1-100)",
    hue="Cluster",
    palette="Set2",
    s=100
)
plt.title("Cluster Segmentation using K_Means")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.legend(title="Cluster")
plt.show()
```

### Plot Centroids

```python
centroids = scaler.inverse_transform(
    kmeans.cluster_centers_
)
centroids
```

```python
plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=df,
    x="Annual Income (k$)",
    y="Spending Score (1-100)",
    hue="Cluster",
    palette="Set2",
    s=100
)

plt.scatter(
    centroids[:,0],
    centroids[:,1],
    marker="x",
    s=300,
    color="black",
    label="Centroids"
)

plt.title("K-Means cluster Segmentation")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")

plt.legend()
plt.show()
```
